# 03 - Run dbt Pipeline (Silver to Gold)
Chạy lệnh dbt build thông qua Python subprocess để đẩy từ Silver sang Gold Star Schema.

```python
import os
import subprocess
import sys

# Chuyển hướng thư mục làm việc (CWD) tới dbt_project
dbt_project_dir = os.path.abspath("../dbt_project")
os.chdir(dbt_project_dir)
print(f"📁 CWD: {os.getcwd()}")

# Cài đặt requirements
print("🚀 Cài đặt thư viện dbt-fabric...")
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "../requirements.txt", "--quiet"], check=True)

# Chạy lệnh dbt deps
print("🚀 Chạy dbt deps...")
subprocess.run(["dbt", "deps"], check=True)

# Chạy lệnh dbt build
print("🚀 Chạy dbt build...")
process = subprocess.run(["dbt", "build"], capture_output=False)

if process.returncode != 0:
    print(f"⚠️ dbt build có lỗi (Exit code: {process.returncode}).")
else:
    print("🎉 dbt build hoàn tất thành công!")
```
